```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt


font = {'size' : 10}
mpl.rc('font', **font)
```

```python
def poly2(x, a, b, c):
    return a*x**2 + b*x + c
```

```python
h_zehnder   = np.array([7, 7])
rho_zehnder = np.array([1000, 3000])

h_chisholm   = np.array([6, 7.5])
rho_chisholm = np.array([1400, 1600])

h_dawney = np.array([4.5, 6.5])
rho_dawney = np.array([4500, 6500])

h_devany = np.array([7, 7])
rho_devany = np.array([5000, 10_000])

rho_lastad = np.arange(1000, 3000)
h_lastad = poly2(rho_lastad, -2e-07,  0.0034,  1.35)
V_lastad = poly2(rho_lastad, 1.67750694e-04, -1.24992232e+00,  5.46534373e+03)

plt.figure(figsize=(8,3))
plt.subplot(121)
plt.plot(rho_zehnder,  h_zehnder,       label="Zehnder et al.")
plt.plot(rho_chisholm, h_chisholm, 'o', label="Chisholm et al.")
plt.plot(rho_dawney,   h_dawney,        label="Dawney et al.")
plt.plot(rho_devany,   h_devany,        label="Devany et al.")
plt.plot(rho_lastad,   h_lastad, 'k--', label="Låstad et al.")

plt.xlabel(r"Cell number density (mm$^{-2}$)")
plt.ylabel(r"Cell height (µm)")

plt.subplot(122)
plt.plot(rho_zehnder,  1e6 * h_zehnder / rho_zehnder,        label="Zehnder et al.")
plt.plot(rho_chisholm, 1e6 * h_chisholm / rho_chisholm, 'o', label="Chisholm et al.")
plt.plot(rho_dawney,   1e6 * h_dawney / rho_dawney,          label="Dawney et al.")
plt.plot(rho_devany,   1e6 * h_devany / rho_devany,          label="Devany et al.")
plt.plot(rho_lastad,   V_lastad, 'k--',   label="Låstad et al.")

plt.xlabel(r"Cell number density (mm$^{-2}$)")
plt.ylabel(r"Cell volume (µm$^3$)")

plt.legend(
    
)
plt.tight_layout()
plt.savefig("figs/Elife/literature_cell_size.png", dpi=300)
```
